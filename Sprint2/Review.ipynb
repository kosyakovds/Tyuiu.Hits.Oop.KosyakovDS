```polyglot-notebook
using System;
using System.Collections.Generic;

public interface IReportable
{
    string GenerateReport();
}

public interface ITeamLeader
{
    List<string> GetTeamMembers();
}

public class Employee
{
    public string Name { get; set; }
    public string Position { get; set; }
    public double Salary { get; set; }

    public Employee(string name, string position, double salary)
    {
        Name = name;
        Position = position;
        Salary = salary;
    }

    public virtual void PrintInfo()
    {
        Console.WriteLine($"Сотрудник: {Name}, Должность: {Position}, Зарплата: {Salary}");
    }
}

public class Manager : Employee
{
    public string Department { get; set; }

    public Manager(string name, string position, double salary, string department)
        : base(name, position, salary)
    {
        Department = department;
    }

    public override void PrintInfo()
    {
        base.PrintInfo();
        Console.WriteLine($"Отдел: {Department}");
    }
}

public class ProjectManager : Manager, IReportable, ITeamLeader
{
    public string ProjectName { get; set; }
    private List<string> _teamMembers;

    public ProjectManager(string name, string position, double salary, string department, string projectName)
        : base(name, position, salary, department)
    {
        ProjectName = projectName;
        _teamMembers = new List<string>();
    }

    public override void PrintInfo()
    {
        base.PrintInfo();
        Console.WriteLine($"Проект: {ProjectName}");
    }

    public string GenerateReport()
    {
        return $"ОТЧЕТ ПРОЕКТНОГО МЕНЕДЖЕРА\n" +
               $"Имя: {Name}\n" +
               $"Проект: {ProjectName}\n" +
               $"Отдел: {Department}\n" +
               $"Команда: {_teamMembers.Count} чел.\n" +
               $"Зарплата: {Salary}";
    }

    public List<string> GetTeamMembers()
    {
        return new List<string>(_teamMembers);
    }

    public void AddTeamMember(string memberName)
    {
        if (!_teamMembers.Contains(memberName))
        {
            _teamMembers.Add(memberName);
            Console.WriteLine($"{memberName} добавлен в команду проекта '{ProjectName}'.");
        }
    }
}

var employee = new Employee("Алексей Петров", "Разработчик", 80000);
var manager = new Manager("Мария Иванова", "Руководитель отдела", 120000, "IT-отдел");
var projectManager = new ProjectManager(
    "Дмитрий Сидоров",
    "Старший менеджер",
    150000,
    "Отдел разработки",
    "Яндекс 2.0"
);

projectManager.AddTeamMember("Анна Козлова");
projectManager.AddTeamMember("Сергей Волков");
projectManager.AddTeamMember("Елена Морозова");

employee.PrintInfo();
manager.PrintInfo();
projectManager.PrintInfo();
Console.WriteLine(projectManager.GenerateReport());

Console.WriteLine("Члены команды проекта:");
foreach (var member in projectManager.GetTeamMembers())
{
    Console.WriteLine($"  - {member}");
}
```
