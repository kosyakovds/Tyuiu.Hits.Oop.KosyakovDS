```polyglot-notebook
using System;
using System.Collections.Generic;
using System.Linq;

public interface IInstallable
{
    void Install(Computer computer);
}

public abstract class Component : IInstallable
{
    public string Name { get; set; }
    public string Manufacturer { get; set; }
    public double Price { get; set; }

    protected Component(string name, string manufacturer, double price)
    {
        Name = name;
        Manufacturer = manufacturer;
        Price = price;
    }

    public virtual string GetInfo()
    {
        return $"{Name}, производитель: {Manufacturer}, цена: {Price} руб.";
    }

    public abstract void Install(Computer computer);

    public override string ToString()
    {
        return GetInfo();
    }
}

public class CPU : Component
{
    public int Cores { get; set; }
    public double Frequency { get; set; }

    public CPU(string name, string manufacturer, double price, int cores, double frequency)
        : base(name, manufacturer, price)
    {
        Cores = cores;
        Frequency = frequency;
    }

    public override string GetInfo()
    {
        return $"{base.GetInfo()}, ядер: {Cores}, частота: {Frequency} ГГц";
    }

    public override void Install(Computer computer)
    {
        Console.WriteLine($"Установка процессора {Name}.");
    }

    public static bool operator ==(CPU a, CPU b)
    {
        if (ReferenceEquals(a, b)) return true;
        if (a is null || b is null) return false;
        return a.Name == b.Name && a.Manufacturer == b.Manufacturer;
    }

    public static bool operator !=(CPU a, CPU b) => !(a == b);

    public override bool Equals(object obj) => obj is CPU other && this == other;
    public override int GetHashCode() => HashCode.Combine(Name, Manufacturer);
}

public class RAM : Component
{
    public int Capacity { get; set; }
    public int Frequency { get; set; }

    public RAM(string name, string manufacturer, double price, int capacity, int frequency)
        : base(name, manufacturer, price)
    {
        Capacity = capacity;
        Frequency = frequency;
    }

    public override string GetInfo()
    {
        return $"{base.GetInfo()}, объём: {Capacity} ГБ, частота: {Frequency} МГц";
    }

    public override void Install(Computer computer)
    {
        Console.WriteLine($"Установка оперативной памяти {Name}.");
    }

    public static bool operator ==(RAM a, RAM b)
    {
        if (ReferenceEquals(a, b)) return true;
        if (a is null || b is null) return false;
        return a.Name == b.Name && a.Manufacturer == b.Manufacturer;
    }

    public static bool operator !=(RAM a, RAM b) => !(a == b);

    public override bool Equals(object obj) => obj is RAM other && this == other;
    public override int GetHashCode() => HashCode.Combine(Name, Manufacturer);
}

public class GPU : Component
{
    public int Memory { get; set; }

    public GPU(string name, string manufacturer, double price, int memory)
        : base(name, manufacturer, price)
    {
        Memory = memory;
    }

    public override string GetInfo()
    {
        return $"{base.GetInfo()}, видеопамять: {Memory} ГБ";
    }

    public override void Install(Computer computer)
    {
        Console.WriteLine($"Установка видеокарты {Name}.");
    }

    public static bool operator ==(GPU a, GPU b)
    {
        if (ReferenceEquals(a, b)) return true;
        if (a is null || b is null) return false;
        return a.Name == b.Name && a.Manufacturer == b.Manufacturer;
    }

    public static bool operator !=(GPU a, GPU b) => !(a == b);

    public override bool Equals(object obj) => obj is GPU other && this == other;
    public override int GetHashCode() => HashCode.Combine(Name, Manufacturer);
}

public class HDD : Component
{
    public int Capacity { get; set; }
    public int RotationSpeed { get; set; }

    public HDD(string name, string manufacturer, double price, int capacity, int rotationSpeed)
        : base(name, manufacturer, price)
    {
        Capacity = capacity;
        RotationSpeed = rotationSpeed;
    }

    public override string GetInfo()
    {
        return $"{base.GetInfo()}, объём: {Capacity} ГБ, скорость вращения: {RotationSpeed} об/мин";
    }

    public override void Install(Computer computer)
    {
        Console.WriteLine($"Установка накопителя {Name}.");
    }

    public static bool operator ==(HDD a, HDD b)
    {
        if (ReferenceEquals(a, b)) return true;
        if (a is null || b is null) return false;
        return a.Name == b.Name && a.Manufacturer == b.Manufacturer;
    }

    public static bool operator !=(HDD a, HDD b) => !(a == b);

    public override bool Equals(object obj) => obj is HDD other && this == other;
    public override int GetHashCode() => HashCode.Combine(Name, Manufacturer);
}

public class Computer
{
    private readonly List<Component> components = new List<Component>();

    public string Name { get; set; }

    public Computer(string name)
    {
        Name = name;
    }

    public void AddComponent<T>(T component) where T : Component, IInstallable
    {
        if (component == null)
        {
            Console.WriteLine("Нельзя добавить пустой компонент.");
            return;
        }

        if (components.Contains(component))
        {
            Console.WriteLine($"Компонент {component.Name} уже добавлен.");
            return;
        }

        component.Install(this);
        components.Add(component);
        Console.WriteLine($"Компонент {component.Name} добавлен в компьютер {Name}.");
    }

    public void RemoveComponent<T>(T component) where T : Component
    {
        if (components.Remove(component))
            Console.WriteLine($"Компонент {component.Name} удалён из компьютера {Name}.");
        else
            Console.WriteLine("Компонент не найден.");
    }

    public double GetTotalPrice()
    {
        return components.Sum(component => component.Price);
    }

    public void ShowComponents()
    {
        Console.WriteLine($"\nКомпоненты компьютера «{Name}»:");

        foreach (Component component in components)
            Console.WriteLine(component.GetInfo());
    }

    public void AddComponent<T>(T component, int quantity) where T : Component, IInstallable
    {
        for (int i = 0; i < quantity; i++)
            AddComponent(component);
    }
}

public static class ComputerFactory
{
    public static Computer CreateConfiguration<T>(string name, params T[] components)
        where T : Component, IInstallable
    {
        Computer computer = new Computer(name);

        foreach (T component in components)
            computer.AddComponent(component);

        return computer;
    }

    public static Computer CreateOfficeComputer()
    {
        return CreateConfiguration<Component>(
            "Офисный компьютер",
            new CPU("Core i3", "Intel", 12000, 4, 4.0),
            new RAM("DDR4 8GB", "Kingston", 2500, 8, 3200),
            new HDD("Barracuda", "Seagate", 4000, 1000, 7200)
        );
    }

    public static Computer CreateGamingComputer()
    {
        return CreateConfiguration<Component>(
            "Игровой компьютер",
            new CPU("Ryzen 7", "AMD", 28000, 8, 4.5),
            new RAM("DDR5 32GB", "Kingston", 9000, 32, 5600),
            new GPU("RTX 4070", "NVIDIA", 65000, 12),
            new HDD("Barracuda", "Seagate", 4000, 2000, 7200)
        );
    }
}

CPU cpu = new CPU("Core i5", "Intel", 18000, 6, 4.2);
RAM ram = new RAM("DDR4 16GB", "Kingston", 4500, 16, 3200);
GPU gpu = new GPU("RTX 4060", "NVIDIA", 35000, 8);
HDD hdd = new HDD("Barracuda", "Seagate", 4000, 1000, 7200);

Computer computer = new Computer("Мой компьютер");

computer.AddComponent(cpu);
computer.AddComponent(ram);
computer.AddComponent(gpu);
computer.AddComponent(hdd);

computer.ShowComponents();
Console.WriteLine($"\nОбщая стоимость: {computer.GetTotalPrice()} руб.");

computer.RemoveComponent(ram);

CPU anotherCPU = new CPU("Core i5", "Intel", 18000, 6, 4.2);
Console.WriteLine($"\nСравнение процессоров: {cpu == anotherCPU}");

Computer office = ComputerFactory.CreateOfficeComputer();
office.ShowComponents();
Console.WriteLine($"Стоимость офисного компьютера: {office.GetTotalPrice()} руб.");

Computer gaming = ComputerFactory.CreateGamingComputer();
gaming.ShowComponents();
Console.WriteLine($"Стоимость игрового компьютера: {gaming.GetTotalPrice()} руб.");
```
